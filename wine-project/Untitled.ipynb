```python
import numpy as np

# Example with a single float
value = 3.7
rounded_value = np.rint(value)
print(f"Rounding {value} to nearest integer: {rounded_value}")

# Example with a NumPy array
arr = np.array([-0.7, 1.5, 2.3, 4.5])
rounded_arr = np.rint(arr)
print(f"Rounding array {arr} to nearest integer: {rounded_arr}")
```
