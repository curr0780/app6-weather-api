# Load the dataframe

```python
import pandas as pd
df = pd.read_csv("wines.csv")
df
```

# How many wines have been given a rating of 100 points?

```python
len(df.loc[df['points'] == 100])
```

# What is the name of the most expensive wine?

```python
from pprint import pprint
expensive = df.loc[df['price'] == df['price'].max()]
#pprint(expensive.squeeze().to_dict())
print(expensive['name'].values)
```

# Calculate a new column where you show the ratings in a scale from 0 to 5. Floats are not allowed.

```python
import numpy as np
df['new rating'] = (df['points'] * 5 / 100)
df['new rating'] = np.rint(df['new rating'])
df.columns
df['new rating']
#df['new rating'].max()
```
